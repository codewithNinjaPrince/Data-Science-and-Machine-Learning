```python
#  Scenario: University Library Search
# Imagine you’re building a smart search system for a university library.
# - Problem: Students often type queries like “fast dog running” or “canine sprinting”, but
# the library catalog has books titled “The dog runs fast”. Traditional keyword search misses these connections.
# - Solution: Use embeddings to capture semantic meaning instead of exact words.


from sentence_transformers import SentenceTransformer

# Choose from popular models:
# "all-MiniLM-L6-v2" → 384 dims, lightweight, fast
# "all-mpnet-base-v2" → 768 dims, stronger semantic understanding
# "BAAI/bge-m3" → 1024 dims, multilingual + dense/sparse retrieval
# "sentence-transformers/e5-large-v2" → 1024 dims, top performer on MTEB

model_name = "all-MiniLM-L6-v2"   # swap this to test other models
model = SentenceTransformer(model_name)

texts = [
    "The dog runs fast",
    "A quick canine sprints",
    "Python is a programming language"
]

embeddings = model.encode(texts)

print(f"Model: {model_name}")
print(f"Embedding shape: {embeddings.shape}")   # e.g. (3, 384) for MiniLM
print(f"First vector sample: {embeddings[0][:10]}")  # show first 10 values
```

```python
# Scenario: Hospital Patient Record Search
# Problem: Doctors and nurses often type queries like “heart attack” or “chest pain emergency”,
# but the hospital’s electronic health record (EHR) system stores diagnoses as “acute myocardial infarction”.
# Traditional keyword search fails because medical staff use everyday language while records use clinical terminology.
# Solution: Use embeddings to capture semantic meaning.
# Embeddings understand that “heart attack” and “acute myocardial infarction” are related concepts.
# This allows the search system to return the right patient records even when the query wording doesn’t match the stored terminology.

from sentence_transformers import SentenceTransformer
# Load embedding model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Patient records stored in hospital database
records = [
    "Acute myocardial infarction",
    "Hypertension",
    "Cardiac arrest",
    "Diabetes mellitus"
]

# Convert records into embeddings (vectors)
record_vectors = model.encode(records)

# Doctor query
query = "heart attack"

# Convert query into embedding
query_vector = model.encode([query])

# Calculate similarity between query and records
similarities = model.similarity(query_vector, record_vectors)

# Print results
for record, score in zip(records, similarities[0]):
    print(f"{record} -> Similarity Score: {score:.3f}")

# Find best match
best_index = similarities.argmax()
print("\nMost Relevant Record:", records[best_index])
```

```python
# Scenario: E‑Commerce Product Search
# - Problem: Shoppers often type queries like “cheap running shoes” or “affordable sneakers for jogging”,
# but the catalog lists products as “Budget-friendly athletic footwear”.
# Traditional keyword search misses these connections because customers use everyday language while product listings use marketing terms.
# - Solution: Use embeddings to capture semantic meaning. Embeddings understand that “cheap running shoes” and “budget-friendly athletic footwear” are related concepts.
# This ensures customers find the right products even when their wording doesn’t match the catalog descriptions.

from sentence_transformers import SentenceTransformer
# Load embedding model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Patient records stored in hospital database
records = [
    "cattle",
    "fridge",
    "washing machine",
    "bulb"
]

# Convert records into embeddings (vectors)
record_vectors = model.encode(records)

# Doctor query
query1 = "light"
query2="clean clothes"

# Convert query into embedding
query_vector1 = model.encode([query1])
query_vector2 = model.encode([query2])

# Calculate similarity between query and records
similarities1 = model.similarity(query_vector1, record_vectors)
similarities2 = model.similarity(query_vector2, record_vectors)

# Print results for query 1
for record, score in zip(records, similarities1[0]):
    print(f"{record} -> Similarity Score: {score:.3f}")
# Print results for query 2
for record, score in zip(records, similarities2[0]):
    print(f"{record} -> Similarity Score: {score:.3f}")

# Find best match
print(f'this is for query 1:{query1}')
best_index = similarities1.argmax()
print("Most Relevant Record:", records[best_index])


# Find best match
print(f'\nthis is for query 2:{query2}')
best_index = similarities2.argmax()
print("Most Relevant Record:", records[best_index])
```

```python
# 🛒 Scenario: E‑Commerce Product Recommendations
# Imagine you’re building a smart recommendation engine for an online store.
# - Problem: Customers often search with phrases like “comfortable running shoes”, but the catalog has items
#  titled “lightweight sneakers for jogging”. Traditional keyword search struggles to connect these.
# - Solution: Use embeddings to capture semantic meaning of product descriptions and customer queries.

from sentence_transformers import SentenceTransformer

# Load embedding model
model = SentenceTransformer('all-MiniLM-L6-v2')


# Product catalog stored in database
products = [
    "lightweight sneakers for jogging",
    "budget friendly running shoes",
    "comfortable cotton t shirt",
    "wireless bluetooth headphones",
    "sports running shorts"
]


# Convert products into embeddings
product_vectors = model.encode(products)


# Customer queries
query1 = "comfortable running shoes"
query2 = "cheap sneakers for exercise"


# Convert queries into embeddings
query_vector1 = model.encode([query1])
query_vector2 = model.encode([query2])


# Calculate similarity
similarities1 = model.similarity(query_vector1, product_vectors)
similarities2 = model.similarity(query_vector2, product_vectors)


# Print results for query 1
print("Results for query 1:\n")

for product, score in zip(products, similarities1[0]):
    print(f"{product} -> Similarity Score: {score:.3f}")


# Print results for query 2
print("\nResults for query 2:\n")

for product, score in zip(products, similarities2[0]):
    print(f"{product} -> Similarity Score: {score:.3f}")


# Best match for query 1
print(f"\nBest match for query 1: {query1}")

best_index = similarities1.argmax()

print("Recommended Product:", products[best_index])


# Best match for query 2
print(f"\nBest match for query 2: {query2}")

best_index = similarities2.argmax()

print("Recommended Product:", products[best_index])
```

```python
# Scenario: Hospital FAQ Assistant
# A hospital wants to build a chatbot that helps patients quickly find answers to common questions.
# - Problem: Patients ask questions in different ways. For example:
# - “How do I book an appointment?”
# - “What’s the process to schedule a doctor visit?”
# - “Can I see a physician tomorrow?”
# Traditional keyword search might miss these connections.
# - Solution: Convert FAQs and patient queries into embeddings, then measure similarity to find the closest match.

# 👩‍💻 Code in Context
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# -----------------------------
# Step 1: Pretend these are FAQ embeddings
# (In reality, you'd generate them with a model like MiniLM or mpnet)
# Each vector is a simplified 4-dimensional embedding
faq1 = np.array([0.2, 0.8, -0.3, 0.5])   # "How do I book an appointment?"
faq2 = np.array([0.1, 0.9, -0.2, 0.4])   # "Process to schedule a doctor visit"
faq3 = np.array([-0.7, 0.1, 0.9, -0.5])  # "Visiting hours for patients"

# -----------------------------
# Step 2: Student query embedding
# Again, simplified for demonstration
query = np.array([0.19, 0.82, -0.25, 0.48])  # "Can I see a physician tomorrow?"

# -----------------------------
# Step 3: Cosine similarity
# Measures the angle between vectors (direction matters, not magnitude)
def cosine_sim(x, y):
    return np.dot(x, y) / (np.linalg.norm(x) * np.linalg.norm(y))

cos_sim1 = cosine_sim(query, faq1)
cos_sim2 = cosine_sim(query, faq2)
cos_sim3 = cosine_sim(query, faq3)

print("Cosine similarities (higher = more similar):")
print("FAQ1:", cos_sim1)
print("FAQ2:", cos_sim2)
print("FAQ3:", cos_sim3)

# -----------------------------
# Step 4: Euclidean distance
# Measures actual geometric distance (lower = more similar)
dist1 = np.linalg.norm(query - faq1)
dist2 = np.linalg.norm(query - faq2)
dist3 = np.linalg.norm(query - faq3)

print("\nEuclidean distances (lower = more similar):")
print("FAQ1:", dist1)
print("FAQ2:", dist2)
print("FAQ3:", dist3)

# -----------------------------
# Step 5: Dot product (normalized)
# Fastest metric, equivalent to cosine if vectors are normalized
query_norm = query / np.linalg.norm(query)
faq1_norm = faq1 / np.linalg.norm(faq1)
faq2_norm = faq2 / np.linalg.norm(faq2)
faq3_norm = faq3 / np.linalg.norm(faq3)

dot1 = np.dot(query_norm, faq1_norm)
dot2 = np.dot(query_norm, faq2_norm)
dot3 = np.dot(query_norm, faq3_norm)

print("\nDot products (normalized, same as cosine):")
print("FAQ1:", dot1)
print("FAQ2:", dot2)
print("FAQ3:", dot3)

# 📊 What Happens
# - Cosine similarity will show that the query is closest to FAQ1 and FAQ2 (both about appointments).
# - Euclidean distance will also confirm closeness, but it’s more sensitive to vector magnitude.
# - Dot product (if normalized) would give the same ranking as cosine, but faster.

# 🎯 Teaching Point
# This scenario shows how similarity metrics decide which FAQ answer to return:
# - Cosine similarity → best for text queries (robust to phrasing differences).
# - Euclidean distance → useful if absolute position matters (e.g., clustering patient queries).
# - Dot product → fastest option when vectors are normalized, ideal for large‑scale systems.

# 👉 In short: the code isn’t just math — it’s the engine behind semantic search and retrieval systems that make chatbots, recommendation engines, and assistants truly useful.
# Would you like me to design a business scenario (like customer support or HR policy search) next, so you can use it directly in your corporate training sessions?
```
