```python
# A school stores students scores in three subjects
scores = np.array([[85, 78, 92],
                   [88, 74, 90],
                   [70, 65, 80],
                   [95, 89, 96]])

# 1️⃣ Top two students (first two rows)
totals = scores.sum(axis=1)
top_two_students = scores[np.argsort(-totals)[:2]]
top_two_students

# 2️⃣ All Math scores (first column)
math_scores = scores[:, 0]
math_scores

# 3️⃣ Students who scored > 90 in Science (third column)
high_science_students = scores[:, 2] > 90 #array([ True, False, False,  True])
high_science_students = scores[high_science_students]
# array([[85, 78, 92],
#        [95, 89, 96]])
high_science_students
```

```python
# Scenario: Farm Crop Yield Analysis
# A farm tracks the yield (in kilograms) of three different crops — Wheat, Rice, and Corn — across four seasons (Spring, Summer, Autumn, Winter).
# The farm manager wants to analyze the data to make better decisions for next year.
# Tasks
# - Create a 2D array representing crop yields for 3 crops across 4 seasons.
# - Rows = crops
# - Columns = seasons
# - Increase all yields by 10% to account for improved irrigation.
# - Extract only Wheat’s yields (the first crop).
# - Find the average yield per crop across all seasons.
# - Identify yields greater than 500 kg (to see which crops/seasons performed exceptionally well).

yields=np.array([[300,500,600,400], #Wheat in 4 Season
         [300,400,500,800], # Rice in 4 season
         [900,400,500,800]]) # Corn in 4 Season
Increased_yields=yields*1.10
Increased_yields
# array([[330., 550., 660., 440.],
#        [330., 440., 550., 880.],
#        [990., 440., 550., 880.]])

wheat_yeilds=yields[0]
wheat_yeilds # array([300, 500, 600, 400])
yields.mean() # np.float64(533.3333333333334)
Yields_greater_than_500=yields[yields>500]
Yields_greater_than_500
```

```python
# Rows = Products (Gadgets, Tools, Machines)
# Columns = Quarters (Q1, Q2, Q3, Q4)

production = np.array([[450, 520, 480, 500],   # Gadgets
                       [600, 580, 620, 610],   # Tools
                       [700, 720, 690, 710]])  # Machines

# Increase all production by 5%
production = production * 1.05

# Extract Gadgets production (first row)
gadgets_production = production[0]

# Average production per product (row-wise mean)
average_per_product = production.mean(axis=1)

# Values greater than 500 units
above_500 = production[production > 500]

print("Updated Production:\n", production)
print("\nGadgets Production:\n", gadgets_production)
print("\nAverage Production per Product:\n", average_per_product)
print("\nProduction values > 500:\n", above_500)
```

```python
# Scenario: Corporate Employee Performance Dashboard
# A company tracks quarterly performance scores of employees across different departments.
# The HR analytics team wants to analyze this dataset to identify top performers, departmental trends, and highlight exceptional scores.

# Dataset
# The dataset contains the following columns:
# - Employee: Name of the employee
# - Department: Department they belong to (Sales, HR, IT, Finance, etc.)
# - Quarter: The quarter in which the performance score was recorded (Q1, Q2, Q3, Q4)
# - Score: Performance score (numeric value)

# Tasks
# - Filtering (Core Corporate Skill)
# - Find employees with Score > 90 (exceptional performers).
# - Filter records for the Sales department only.
# - Sorting (Power Moment)
# - Sort employees by Score in descending order to rank performance.
# - Sort by Department first, then Score to see departmental rankings.
# - Grouping (Core Corporate Skill)
# - Group by Department and calculate the average score per department.
# - Group by Quarter and find the maximum score achieved in each quarter.

# Business Context
# - Filtering helps HR identify exceptional performers and focus on specific departments.
# - Sorting creates clear performance rankings, useful for promotions or recognition programs.
# - Grouping reveals departmental strengths and quarterly performance trends, guiding training and resource allocation.

import pandas as pd

# 1️⃣ Load Dataset
df = pd.read_csv("Dataset.csv")
df.head()
df.info()
df.describe()

# OUTPUT

# <class 'pandas.core.frame.DataFrame'>
# RangeIndex: 10 entries, 0 to 9
# Data columns (total 4 columns):
#  #   Column      Non-Null Count  Dtype
# ---  ------      --------------  -----
#  0   Employee    10 non-null     object
#  1   Department  10 non-null     object
#  2   Quarter     10 non-null     object
#  3   Score       10 non-null     int64
# dtypes: int64(1), object(3)
# memory usage: 452.0+ bytes
# Score
# count	10.000000
# mean	85.300000
# std	9.177872
# min	70.000000
# 25%	78.500000
# 50%	86.500000
# 75%	91.500000
# max	99.000000


# ==============================
# 🔎 FILTERING (Core Corporate Skill)
# ==============================

# Exceptional performers (Score > 90)
exceptional_performers = df[df["Score"] > 90]
print("\nExceptional Performers (Score > 90):\n", exceptional_performers)

# 	Employee	Department	Quarter	Score
# 2	John	IT	Q1	92
# 4	David	Sales	Q2	95
# 8	Neha	Sales	Q3	99


# Sales Department records only
sales_department = df[df["Department"] == "Sales"]
print("\nSales Department Records:\n", sales_department)

# 	Employee	Department	Quarter	Score
# 0	Amit	Sales	Q1	78
# 4	David	Sales	Q2	95
# 8	Neha	Sales	Q3	99


# ==============================
# 📊 SORTING (Power Moment)
# ==============================

# Sort by Score (Descending)
sorted_by_score = df.sort_values("Score", ascending=False)
print("\nEmployees Ranked by Score:\n", sorted_by_score)

# 	Employee	Department	Quarter	Score
# 6	Alex	IT	Q2	70
# 9	Karan	IT	Q3	76
# 0	Amit	Sales	Q1	78
# 5	Meera	HR	Q2	80
# 1	Priya	HR	Q1	85
# 3	Sara	Finance	Q1	88
# 7	Ravi	Finance	Q2	90
# 2	John	IT	Q1	92
# 4	David	Sales	Q2	95
# 8	Neha	Sales	Q3	99


# Sort by Department first, then Score (Descending)
dept_wise_ranking = df.sort_values(["Department", "Score"], ascending=[True, False])
print("\nDepartment-wise Ranking:\n", dept_wise_ranking)

# 	Employee	Department	Quarter	Score
# 7	Ravi	Finance	Q2	90
# 3	Sara	Finance	Q1	88
# 1	Priya	HR	Q1	85
# 5	Meera	HR	Q2	80
# 2	John	IT	Q1	92
# 9	Karan	IT	Q3	76
# 6	Alex	IT	Q2	70
# 8	Neha	Sales	Q3	99
# 4	David	Sales	Q2	95
# 0	Amit	Sales	Q1	78


# ==============================
# 📈 GROUPING (Core Corporate Skill)
# ==============================

# Average score per Department
avg_score_department = df.groupby("Department")["Score"].mean()
print("\nAverage Score per Department:\n", avg_score_department)

# Department
# Finance    89.000000
# HR         82.500000
# IT         79.333333
# Sales      90.666667
# Name: Score, dtype: float64

# Maximum score per Quarter
max_score_quarter = df.groupby("Quarter")["Score"].max()
print("\nMaximum Score per Quarter:\n", max_score_quarter)

# Quarter
# Q1    92
# Q2    95
# Q3    99
# Name: Score, dtype: int64

# 📌 Business Interpretation (Corporate Understanding)
# 🔹 Filtering

# Identifies top performers (Score > 90).
# Helps HR focus on Sales department performance separately.

# 🔹 Sorting

# Creates performance ranking (important for promotions & bonuses).
# Department-wise sorting shows internal competition & leaders.

# 🔹 Grouping

# Department average score → shows which department performs best overall.
# Quarterly maximum score → identifies peak performance periods.
```
