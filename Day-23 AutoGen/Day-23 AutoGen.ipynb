```python
# Scenario: AI Code Review & Execution System (Corporate Dev Team)
# 🎯 Problem Statement

# In companies:

# Developers write code

# Reviewers check for bugs

# Testing is manual

# 👉 Solution:
# A multi-agent system where:

# 👨‍💻 Coder Agent writes code

# 🔍 Reviewer Agent validates + executes code

# 🔁 Loop continues until correct


# -----------------------------
# Imports
# -----------------------------
import os
import autogen
from google.colab import userdata

# -----------------------------
# Groq API Key
# -----------------------------
groq_api_key = userdata.get("GROQ_API_KEY")
os.environ["GROQ_API_KEY"] = groq_api_key

# -----------------------------
# LLM Config (Groq)
# -----------------------------
llm_config = {
    "config_list": [
        {
            "model": "llama-3.3-70b-versatile",
            "api_key": groq_api_key,
            "base_url": "https://api.groq.com/openai/v1"
        }
    ],
    "temperature": 0
}

# -----------------------------
# Agent 1: Coder
# -----------------------------
coder = autogen.AssistantAgent(
    name="Coder",
    llm_config=llm_config,
    system_message="""You are an expert Python developer.

Write clean, well-commented code.

Rules:
- Handle edge cases
- Add test cases
- Ensure code runs without errors
- End your response with DONE"""
)

# -----------------------------
# Agent 2: Reviewer (Executor)
# -----------------------------
reviewer = autogen.UserProxyAgent(
    name="Reviewer",
    human_input_mode="NEVER",
    code_execution_config={
        "work_dir": "review",
        "use_docker": False
    },
    is_termination_msg=lambda x: "DONE" in x.get("content", ""),
    system_message="""You are a code reviewer.

Responsibilities:
1. Review the code for bugs
2. Execute the code
3. Validate outputs
4. Ask for fixes if incorrect
5. Approve when correct"""
)

# -----------------------------
# Start Collaboration
# -----------------------------
reviewer.initiate_chat(
    coder,
    message="""Create a Python function that:
- Generates Fibonacci sequence up to N
- Returns it as a list
- Handles edge cases (N=0, negative N)

Test it with N=10 and print output"""
)
```

```python
# Scenario: AI Marketing Content & Validation System (Corporate Marketing Team)
# 🎯 Problem Statement
# In companies:
# - Marketing teams draft campaign content manually.
# - Editors check for tone, compliance, and brand alignment.
# - Testing across platforms (emails, social posts, ads) is manual and time-consuming.
# 👉 Solution
# A multi-agent system where:
# - ✍️ Content Creator Agent generates marketing copy, visuals, or ad text.
# - 🔎 Validator Agent reviews for compliance, tone, and brand guidelines, then simulates deployment.
# - 🔁 The loop continues until the content meets quality and compliance standards.
# 🔄 Workflow
# Campaign Request → AI drafts content → Validator checks tone/compliance → Test deployment simulation → Iteration loop → Final approved campaign assets
# 🧹 Conflict Resolution
# - Resolve conflicting brand guidelines (e.g., tone mismatch between product lines).
# - Standardize formatting across channels (social, email, ads).
# - Ensure package/library consistency for creative assets (e.g., Canva API vs. Adobe API).

# =====================================================
# Scenario: AI Marketing Content & Validation System
# =====================================================

# Problem:
# Marketing team writes content
# Editor validates tone / compliance
# Deployment test is manual
#
# Solution:
# Multi-agent loop:
# ContentCreator -> Validator -> Fix -> Validate -> Approve


# -----------------------------
# Imports
# -----------------------------
import os
import autogen
from google.colab import userdata


# -----------------------------
# Groq API Key
# -----------------------------
groq_api_key = userdata.get("GROQ_API_KEY")
os.environ["GROQ_API_KEY"] = groq_api_key


# -----------------------------
# LLM Config (Groq)
# -----------------------------
llm_config = {
    "config_list": [
        {
            "model": "llama-3.3-70b-versatile",
            "api_key": groq_api_key,
            "base_url": "https://api.groq.com/openai/v1"
        }
    ],
    "temperature": 0.3
}


# =====================================================
# Agent 1: Content Creator
# =====================================================

creator = autogen.AssistantAgent(
    name="ContentCreator",
    llm_config=llm_config,
    system_message="""
You are a senior marketing copywriter.

Your job:
- Create campaign content
- Follow brand tone: professional + friendly
- Ensure compliance
- Provide content for:
  - Email
  - Social media
  - Advertisement

Rules:
- Keep tone consistent
- Avoid prohibited claims
- Format clearly
- End response with DONE
"""
)


# =====================================================
# Agent 2: Validator + Executor
# =====================================================

validator = autogen.UserProxyAgent(
    name="Validator",
    human_input_mode="NEVER",

    code_execution_config={
        "work_dir": "marketing_review",
        "use_docker": False
    },

    is_termination_msg=lambda x: "DONE" in x.get("content", ""),

    system_message="""
You are a marketing compliance reviewer.

Responsibilities:

1. Check tone consistency
2. Check brand guidelines
3. Check legal compliance
4. Simulate deployment
5. Ask for fixes if incorrect
6. Approve only if perfect

Brand Rules:
- Tone must be professional + friendly
- No false claims
- No medical / financial promises
- Format must include:
    Email
    Social Post
    Ad Copy

If wrong → ask for correction
If correct → say APPROVED DONE
"""
)


# =====================================================
# Start Workflow
# =====================================================

validator.initiate_chat(
    creator,
    message="""
Create a marketing campaign for:

Product: AI Project Management Tool
Audience: Corporate teams
Tone: Professional + Friendly

Required outputs:
- Email campaign
- LinkedIn post
- Ad copy

Ensure compliance and brand tone.
Simulate deployment readiness.
"""
)
```

```python
# Scenario: AI-Powered Dev Workflow (Corporate Engineering Team)
# 🎯 Problem Statement
# In companies:
# - Developers manually write and test code.
# - Reviewers check for bugs and compliance.
# - Testing and validation are slow and error-prone.
# 👉 Solution
# A multi-agent system powered by Groq API where:
# - 👨‍💻 Coder Agent writes Python code for requested features.
# - 🔍 Reviewer Agent validates logic and ensures best practices.
# - 🧪 Tester Agent generates unit tests and executes them automatically.
# - 🔁 The loop continues until the code passes all tests and meets requirements.
# 🔄 Workflow
# Feature Request → Product Manager breaks into tasks → Coder writes code → Tester validates with unit tests → Reviewer checks compliance → Iteration loop → Final approved codebase
# 🧹 Conflict Resolution
# - Resolve conflicting dependencies (e.g., package version mismatches).
# - Standardize coding style across agents (PEP8 compliance).
# - Ensure Groq API integration avoids latency or model mismatch issues.


import autogen
import os
from google.colab import userdata

# Groq API configuration
# Make sure you have stored your Groq API key in Colab Secrets as 'GROQ_API_KEY'
groq_api_key = userdata.get("GROQ_API_KEY")

llm_config = {
    "config_list": [
        {
            "model": "llama-3.3-70b-versatile",   # Updated to a supported Groq model
            "api_key": groq_api_key,
            "base_url": "https://api.groq.com/openai/v1"  # Groq endpoint
        }
    ]
}

# Specialized agents
pm = autogen.AssistantAgent(
    name="ProductManager",
    llm_config=llm_config,
    system_message="You break down feature requests into tasks."
)

coder = autogen.AssistantAgent(
    name="Coder",
    llm_config=llm_config,
    system_message="You write Python code for given tasks."
)

tester = autogen.AssistantAgent(
    name="Tester",
    llm_config=llm_config,
    system_message="You write unit tests for code."
)

user = autogen.UserProxyAgent(
    name="User",
    human_input_mode="NEVER",
    code_execution_config={"work_dir": "project"}
)

# Group chat setup
groupchat = autogen.GroupChat(
    agents=[user, pm, coder, tester],
    messages=[],
    max_round=12,
    speaker_selection_method="auto"
)

# Manager orchestrates workflow
manager = autogen.GroupChatManager(
    groupchat=groupchat,
    llm_config=llm_config
)

# Start the group task
user.initiate_chat(manager,
    message="Build a simple REST API for a todo app")
```

```python
# Scenario: AI-Powered HR Onboarding System (Corporate HR Team)
# 🎯 Problem Statement
# In companies:
# - HR teams manually prepare onboarding documents and checklists.
# - Managers validate compliance with policies.
# - Employees often face delays in receiving resources and training schedules.
# 👉 Solution
# A multi-agent system where:
# - 📝 HR Agent prepares onboarding documents, training schedules, and resource lists.
# - ✅ Compliance Agent reviews documents for policy adherence and legal requirements.
# - 📂 Resource Agent allocates accounts, devices, and access permissions.
# - 🔁 The loop continues until all onboarding steps are validated and complete.
# 🔄 Workflow
# Onboarding Request → HR Agent drafts documents → Compliance Agent validates → Resource Agent provisions accounts/devices → Iteration loop → Final approved onboarding package
# 🧹 Conflict Resolution
# - Resolve conflicting policy requirements (e.g., different regional compliance rules).
# - Standardize onboarding templates across departments.
# - Ensure system avoids duplicate resource allocation (e.g., multiple accounts for same employee).


import autogen
import os
from google.colab import userdata


# -------------------------------------------------
# Groq API configuration
# -------------------------------------------------

groq_api_key = userdata.get("GROQ_API_KEY")

llm_config = {
    "config_list": [
        {
            "model": "llama-3.3-70b-versatile",
            "api_key": groq_api_key,
            "base_url": "https://api.groq.com/openai/v1"
        }
    ],
    "temperature": 0.2
}


# =====================================================
# Agent 1 — HR Manager (Task Planner)
# =====================================================

hr_manager = autogen.AssistantAgent(
    name="HRManager",
    llm_config=llm_config,
    system_message="""
You manage onboarding workflow.

Responsibilities:
- Break onboarding request into steps
- Assign tasks to HR / Compliance / Resource agents
- Ensure final onboarding package complete
"""
)


# =====================================================
# Agent 2 — HR Agent
# =====================================================

hr_agent = autogen.AssistantAgent(
    name="HRAgent",
    llm_config=llm_config,
    system_message="""
You prepare onboarding materials.

Create:
- Offer letter
- Onboarding checklist
- Training schedule
- Welcome email

Rules:
- Use company template
- Avoid missing fields
- Standardize format
"""
)


# =====================================================
# Agent 3 — Compliance Agent
# =====================================================

compliance_agent = autogen.AssistantAgent(
    name="ComplianceAgent",
    llm_config=llm_config,
    system_message="""
You verify HR documents.

Check:
- Policy compliance
- Legal requirements
- Regional rules
- Duplicate data
- Template correctness

If incorrect → request fix
If correct → approve
"""
)


# =====================================================
# Agent 4 — Resource Agent
# =====================================================

resource_agent = autogen.AssistantAgent(
    name="ResourceAgent",
    llm_config=llm_config,
    system_message="""
You allocate company resources.

Provide:
- Email account
- Laptop
- System access
- Training portal access

Rules:
- No duplicate accounts
- Follow department policy
- Match employee role
"""
)


# =====================================================
# User / Executor Agent
# =====================================================

user = autogen.UserProxyAgent(
    name="User",
    human_input_mode="NEVER",
    code_execution_config={
        "work_dir": "onboarding_system",
        "use_docker": False
    }
)


# =====================================================
# Group Chat Setup
# =====================================================

groupchat = autogen.GroupChat(
    agents=[
        user,
        hr_manager,
        hr_agent,
        compliance_agent,
        resource_agent
    ],
    messages=[],
    max_round=15,
    speaker_selection_method="auto"
)


manager = autogen.GroupChatManager(
    groupchat=groupchat,
    llm_config=llm_config
)


# =====================================================
# Start Workflow
# =====================================================

user.initiate_chat(
    manager,
    message="""
Onboard new employee:

Name: Rahul Sharma
Role: Software Engineer
Location: India
Department: Engineering

Prepare full onboarding package
Check compliance
Allocate resources
Finish only when onboarding complete
"""
)
```
