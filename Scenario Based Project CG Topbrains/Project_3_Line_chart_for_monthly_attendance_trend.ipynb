```python
# Question
# 3
# Matplotlib
# Description
#  📘 Scenario Question # A college is tracking student attendance across the first four months of the semester:
# - January: 85 students present on average # - February: 90 students
# - March: 95 students # - April: 88 students
# The academic coordinator wants to visualize the attendance trend to see whether students are attending classes more regularly or
# if there are fluctuations. # You are asked to:
# - Plot a line chart showing the monthly attendance trend. # - Label the axes and add a title to make the chart clear.
# - Interpret the chart: # - Which month had the highest attendance? # - Did attendance steadily increase, or were there ups and downs?
# - What might explain the dip in April?

import matplotlib.pyplot as plt


months = ["January", "February", "March", "April"]
attendance = [85, 90, 95, 88]


plt.plot(months, attendance, marker='o', linestyle='-', color='blue')

plt.xlabel("Month")
plt.ylabel("Average Student Attendance")
plt.title("Monthly Student Attendance Trend")


plt.grid(True)

plt.show()


# ANSWERS:-
# 1.March had the highest attendance with 95 students on average.

# 2.Attendance increased from January to March:
# January → 85
# February → 90
# March → 95
# but there was a drop in April to 88, so attendance did not increase steadily throughout.

# 3.there was an increase in the trend initially, followed by a decline in April.
```
